```python
df1 = pd.read_csv('data_withdate.csv')
print(df1.to_string())
```

```python
df2 = df1.dropna()
print(df2)
```

```python
df1.fillna({"Calories":130}, inplace = True)
print(df1)
```

```python
x = df1["Pulse"].mean()
df1.fillna({"Pulse":x}, inplace = True)
print(df1)
```

```python
y = df1["Maxpulse"].mode()[0]
df1.fillna({"Maxpulse":y}, inplace = True)
print(df1)
```

```python
df1["Date"] = pd.to_datetime(df1["Date"], format='mixed')
print(df1)
```

```python
for x in df1.index:
    if df1.loc[x, "Duration"] > 120:
        df1.loc[x, "Duration"] = 120
print(df1.to_string())
```

```python
print(df1.duplicated().to_string())
df1.drop_duplicates(inplace = True)
print(df1.to_string())
```

```python
import pandas as pd

df4 = pd.read_csv('data_short.csv')

print(df4.to_string())
```

```python
df_dropped = df4.dropna()
print(df_dropped)
```

```python
df_filled = df4.fillna(444)
print(df_filled)
```

```python
mean = df4['Calories'].mean()
calories_fixed = df4.fillna({'Calories':mean})
print(calories_fixed)
```

```python
median = df4['Maxpulse'].median()
maxpulse_fixed = df4.fillna({'Maxpulse':median})
print(maxpulse_fixed)
```

```python
mode = df4['Pulse'].mode()[0]
pulse_fixed = df4.fillna({'Pulse':mode})
print(pulse_fixed)
```

```python
duration_fixed = df4
for wrongdata in duration_fixed.index:
    if duration_fixed.loc[wrongdata, 'Duration'] > 90:
        duration_fixed.loc[wrongdata, 'Duration'] = 90
print(duration_fixed)
```
